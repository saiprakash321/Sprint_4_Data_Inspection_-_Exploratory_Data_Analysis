# Module 4 - Univariate Analysis

Univariate analysis examines one variable at a time to understand its distribution, central tendency, spread, and important patterns.

This notebook uses the Hotel Booking Demand dataset to analyze numerical and categorical variables using statistical measures and visualizations.

## 1. Setup and Data Load

The required libraries are imported and the Hotel Booking Demand dataset is loaded for univariate analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 2. Data Quality Check

Before performing statistical analysis, it is important to check for missing values, duplicate records, and invalid values.

This helps ensure that the results are not affected by common data-quality issues.

### Implementation

```python
print("Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate Rows:", df.duplicated().sum())
```

### Interpretation

The output shows the columns with missing values and the number of duplicate rows in the dataset.

### Insight

This initial check helps identify data-quality issues that should be considered while interpreting the univariate analysis.

### AI/ML Usage

Data-quality checks help ensure that statistical analysis and later Machine Learning steps are based on reliable data.

## 3. Numerical Variables

Numerical variables contain values that can be measured or used for mathematical calculations.

For the Hotel Booking Demand dataset, numerical variables include lead_time, adults, children, babies, stays_in_weekend_nights, stays_in_week_nights, previous_cancellations, booking_changes, adr, and total_of_special_requests.

These variables can be analyzed using statistical measures such as mean, median, variance, standard deviation, quartiles, skewness, and kurtosis.

### 3.1 Mean

Mean is the average value of a numerical variable. It is calculated by dividing the sum of all values by the number of values.

### Example

For the values 10, 20, and 30:

Mean = (10 + 20 + 30) / 3 = 20

### Real-World Example

In the Hotel Booking Demand dataset, the mean lead time shows the average number of days between the booking date and the arrival date.

### AI/ML Usage

Mean is used to understand the central tendency of numerical features and can be useful for missing-value imputation when appropriate.

### Implementation

```python
mean_value = df["lead_time"].mean()

print("Mean Lead Time:", round(mean_value, 2))
```

### Interpretation

The output shows the average lead time of the hotel bookings.

### Insight

The mean provides an overall idea of how far in advance customers usually make their hotel reservations.

### AI/ML Relevance

Mean helps us understand the typical value of a numerical feature during Exploratory Data Analysis.

### 3.2 Median

Median is the middle value of an ordered dataset. It divides the data into two equal parts.

### Example

For the values 10, 20, and 30, the median is 20.

### Real-World Example

In the Hotel Booking Demand dataset, the median lead time shows the middle booking lead time and is less affected by very large values.

### AI/ML Usage

Median is useful for understanding the center of skewed data and can also be used to handle missing numerical values.

### Implementation

```python
median_value = df["lead_time"].median()

print("Median Lead Time:", median_value)
```

### Interpretation

The output shows the middle lead time when all booking lead times are arranged in order.

### Insight

The median gives a representative value even when some bookings are made much earlier than others.

### AI/ML Relevance

Median is useful during EDA and can be preferred over mean when numerical data contains extreme values.

### 3.3 Mode

Mode is the value that appears most frequently in a dataset.

It is useful for identifying the most common value in a numerical or categorical variable.

### Example

For the values 10, 20, 20, and 30, the mode is 20 because it occurs most frequently.

### Real-World Example

In the Hotel Booking Demand dataset, the mode of adults shows the most common number of adults included in a booking.

### AI/ML Usage

Mode can be used to understand common patterns and is often used to fill missing values in categorical or discrete variables.

### Implementation

```python
mode_value = df["adults"].mode()[0]

print("Most Common Number of Adults:", mode_value)
```

### Interpretation

The output shows the number of adults that occurs most frequently in the booking records.

### Insight

The mode helps identify the most common booking pattern for the selected variable.

### AI/ML Relevance

Mode is useful during EDA for understanding frequent values and can support missing-value handling for suitable variables.

### 3.4 Minimum and Maximum

Minimum is the smallest value in a numerical variable, while Maximum is the largest value.

### Example

For the values 10, 20, and 30, the minimum is 10 and the maximum is 30.

### Real-World Example

In the Hotel Booking Demand dataset, minimum and maximum lead time show the shortest and longest time between booking and arrival.

### AI/ML Usage

Minimum and maximum values help understand the range of a feature and are useful when identifying unusual values and applying feature scaling.

### Implementation

```python
min_value = df["lead_time"].min()
max_value = df["lead_time"].max()

print("Minimum Lead Time:", min_value)
print("Maximum Lead Time:", max_value)
```

### Interpretation

The output shows the smallest and largest lead time values in the dataset.

### Insight

The difference between the minimum and maximum values gives an initial understanding of how widely lead time varies across bookings.

### AI/ML Relevance

Minimum and maximum values help understand feature ranges and can be useful during data preprocessing and feature scaling.

### 3.5 Range

Range is the difference between the maximum and minimum values in a numerical variable.

### Example

For the values 10, 20, and 30:

Range = 30 - 10 = 20

### Real-World Example

In the Hotel Booking Demand dataset, the range of lead time shows the difference between the shortest and longest booking lead times.

### AI/ML Usage

Range helps understand the overall spread of a numerical feature and provides its minimum and maximum boundaries.

### Implementation

```python
range_value = df["lead_time"].max() - df["lead_time"].min()

print("Lead Time Range:", range_value)
```

### Interpretation

The output shows the difference between the maximum and minimum lead time.

### Insight

A larger range indicates that booking lead times vary widely across the dataset.

### AI/ML Relevance

Range helps understand feature spread during Exploratory Data Analysis and can support decisions about scaling or transformation.

### 3.6 Variance

Variance measures how much the values in a dataset spread out from the mean.

A low variance means the values are close to the mean, while a high variance means the values are more spread out.

### Example

For values 10, 20, and 30, variance shows how far these values are spread from their average value.

### Real-World Example

In the Hotel Booking Demand dataset, variance in lead time shows how much the number of days between booking and arrival differs across bookings.

### AI/ML Usage

Variance helps measure feature variability and is important in techniques such as feature selection and Principal Component Analysis (PCA).

### Implementation

```python
variance_value = df["lead_time"].var()

print("Variance of Lead Time:", round(variance_value, 2))
```

### Interpretation

The output shows how much the lead time values vary around their mean.

### Insight

A higher variance indicates greater variation in how far in advance customers make their bookings.

### AI/ML Relevance

Variance helps identify how much information or variation a numerical feature contains before applying Machine Learning techniques.

### 3.7 Standard Deviation

Standard deviation measures how much the values in a dataset differ from the mean.

A smaller standard deviation means the values are closer to the mean, while a larger standard deviation means the values are more spread out.

### Example

For values 10, 20, and 30, standard deviation shows how far the values are spread from their average.

### Real-World Example

In the Hotel Booking Demand dataset, standard deviation of lead time shows how much booking times vary around the average lead time.

### AI/ML Usage

Standard deviation is commonly used to understand data spread and in standardization of numerical features.

### Implementation

```python
std_value = df["lead_time"].std()

print("Standard Deviation of Lead Time:", round(std_value, 2))
```

### Interpretation

The output shows the average spread of lead time values around the mean.

### Insight

A higher standard deviation indicates more variation in how far in advance customers make bookings.

### AI/ML Relevance

Standard deviation helps understand feature variability and is important when standardizing numerical features for Machine Learning.

### 3.8 Quartiles and Percentiles

Quartiles divide ordered data into four equal parts, while percentiles divide the data into 100 parts.

Q1 represents the 25th percentile, Q2 represents the 50th percentile or median, and Q3 represents the 75th percentile.

### Example

For a dataset, Q1 indicates the value below which 25% of the observations fall, while Q3 indicates the value below which 75% of the observations fall.

### Real-World Example

In the Hotel Booking Demand dataset, quartiles of lead time help us understand how booking days are distributed across customers.

### AI/ML Usage

Quartiles and percentiles help understand feature distribution and are also used to calculate the Interquartile Range (IQR) for identifying potential outliers.

### Implementation

```python
quartiles = df["lead_time"].quantile([0.25, 0.50, 0.75])

print(quartiles)
```

### Interpretation

The output shows the 25th, 50th, and 75th percentile values of lead time.

### Insight

These values help us understand how booking lead times are distributed across the dataset.

### AI/ML Relevance

Quartiles and percentiles are useful for understanding numerical features and detecting potential outliers during EDA.

### 3.9 Skewness

Skewness measures the asymmetry of a numerical data distribution around its mean.

Positive skewness indicates a longer right tail, while negative skewness indicates a longer left tail. A value close to zero indicates a more symmetric distribution.

### Example

If most hotel bookings have shorter lead times but a few bookings are made far in advance, the lead time distribution may be positively skewed.

### Real-World Example

In the Hotel Booking Demand dataset, skewness of lead time helps us understand whether booking periods are evenly distributed or concentrated toward one side.

### AI/ML Usage

Skewness helps identify non-symmetric features and can indicate when a transformation may be useful before applying certain Machine Learning algorithms.

### Implementation

```python
skew_value = df["lead_time"].skew()

print("Skewness of Lead Time:", round(skew_value, 2))
```

### Interpretation

The output shows the skewness of the lead time variable.

### Insight

The skewness value helps us understand whether the booking lead time distribution is symmetric or shifted toward one side.

### AI/ML Relevance

Understanding skewness helps in selecting suitable preprocessing and transformation techniques for numerical features.

### 3.10 Kurtosis

Kurtosis measures the shape of the tails of a numerical distribution and helps identify the presence of extreme values.

A higher kurtosis indicates heavier tails, while a lower kurtosis indicates lighter tails.

### Example

If a few hotel bookings have extremely large lead times compared with most bookings, the distribution may show higher kurtosis.

### Real-World Example

In the Hotel Booking Demand dataset, kurtosis of lead time helps us understand whether extreme booking lead times occur more frequently than expected.

### AI/ML Usage

Kurtosis helps identify the presence of extreme values and can support decisions about outlier treatment and data transformation.

### Implementation

```python
kurtosis_value = df["lead_time"].kurt()

print("Kurtosis of Lead Time:", round(kurtosis_value, 2))
```

### Interpretation

The output shows the kurtosis value of the lead time variable.

### Insight

The kurtosis value helps us understand the tendency of the distribution to contain extreme observations.

### AI/ML Relevance

Kurtosis can help identify features affected by extreme values and guide further EDA and preprocessing decisions.

### 4.1 Histogram

A histogram is a graph used to show the distribution of a numerical variable by dividing the values into intervals called bins.

### Example

A histogram of lead time shows how booking days are distributed across different ranges.

### Real-World Example

A hotel can use a histogram to understand whether most customers book rooms shortly before arrival or well in advance.

### AI/ML Usage

Histograms help identify the shape of a feature distribution, concentration of values, and possible unusual patterns during EDA.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.histplot(df["lead_time"], bins=30)

plt.title("Lead Time Distribution")
plt.xlabel("Lead Time")
plt.ylabel("Count")
plt.show()
```

### Interpretation

The histogram shows how frequently different lead-time ranges occur in the dataset.

### Insight

The distribution helps identify the booking periods where observations are more concentrated and whether the feature has a skewed pattern.

### AI/ML Relevance

Understanding the distribution of numerical features helps decide whether transformations or other preprocessing techniques may be required.

### 4.2 KDE / Density Plot

A KDE (Kernel Density Estimation) plot is a smooth curve used to represent the distribution of a numerical variable.

### Example

A KDE plot of lead time helps us understand where booking values are more concentrated.

### Real-World Example

A hotel can use a density plot to understand the common booking periods and identify areas where bookings are more concentrated.

### AI/ML Usage

KDE plots help analyze the shape of numerical distributions and identify patterns such as concentration, skewness, and multiple peaks.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.kdeplot(df["lead_time"].dropna(), fill=True)

plt.title("Lead Time Density Plot")
plt.xlabel("Lead Time")
plt.ylabel("Density")
plt.show()
```

### Interpretation

The density curve shows how the lead time values are distributed, with higher areas representing greater concentration of observations.

### Insight

The shape of the curve helps us understand where most booking lead times are concentrated and whether the distribution is skewed.

### AI/ML Relevance

Density plots help understand feature distributions and can support decisions about transformations and further statistical analysis.

### 4.3 Box Plot

A box plot is a visual representation of the distribution of a numerical variable. It shows the median, quartiles, and possible outliers.

### Example

A box plot of lead time helps us understand the spread of booking periods and identify unusually high or low values.

### Real-World Example

A hotel can use a box plot to identify unusual booking lead times that may require further investigation.

### AI/ML Usage

Box plots are commonly used during EDA to identify potential outliers and understand the spread of numerical features.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["lead_time"])

plt.title("Lead Time Box Plot")
plt.xlabel("Lead Time")
plt.show()
```

### Interpretation

The box plot shows the median, middle 50% of the data, and values that fall outside the typical range.

### Insight

The plot helps identify the overall spread of lead time and any potential extreme observations.

### AI/ML Relevance

Box plots help detect potential outliers before further statistical analysis and Machine Learning preprocessing.

## 5. Categorical Variables

Categorical variables contain values that represent different groups or categories.

In the Hotel Booking Demand dataset, examples include hotel type, meal type, market segment, distribution channel, deposit type, and customer type.

### 5.1 Frequency

Frequency represents the number of times each category occurs in a dataset.

### Example

The frequency of each hotel type shows how many bookings belong to each hotel category.

### Real-World Example

A hotel can use frequency counts to understand which type of hotel receives more bookings.

### AI/ML Usage

Frequency analysis helps understand category distribution and identify categories that may be rare or dominant.

### Implementation

```python
frequency = df["hotel"].value_counts()

print(frequency)
```

### Interpretation

The output shows the number of bookings for each hotel type.

### Insight

The frequency distribution helps identify which hotel category has more bookings in the dataset.

### AI/ML Relevance

Frequency analysis helps understand categorical feature distributions before encoding and Machine Learning.

### 5.2 Unique Values

Unique values are the distinct values present in a categorical variable.

### Example

The hotel column contains different hotel types, such as City Hotel and Resort Hotel.

### Real-World Example

A hotel company can identify the different types of properties included in its booking data.

### AI/ML Usage

Unique-value analysis helps understand the number of categories in a feature and supports the selection of suitable encoding methods.

### Implementation

```python
unique_values = df["hotel"].unique()

print("Unique Hotel Types:", unique_values)
print("Number of Unique Values:", df["hotel"].nunique())
```

### Interpretation

The output shows the distinct hotel types present in the dataset and the total number of unique categories.

### Insight

This helps us understand the variety of categories available in the selected feature.

### AI/ML Relevance

Knowing the number of categories helps determine how a categorical feature should be represented during Machine Learning preprocessing.

### 5.3 Value Counts

Value counts show how many times each unique category appears in a column.

### Example

The hotel column can be used to count how many bookings belong to each hotel type.

### Real-World Example

A hotel company can use value counts to compare the number of bookings received by different hotel types.

### AI/ML Usage

Value counts help understand category distribution and identify dominant or less frequent categories.

### Implementation

```python
value_counts = df["hotel"].value_counts()

print(value_counts)
```

### Interpretation

The output shows the number of records belonging to each hotel category.

### Insight

The counts help identify which hotel type has the highest number of bookings.

### AI/ML Relevance

Value counts are useful for understanding categorical feature distributions before encoding and model development.

### 5.4 Percentage Distribution

Percentage distribution shows the proportion of each category in a dataset.

It is calculated by dividing the frequency of each category by the total number of records and multiplying by 100.

### Example

The percentage distribution of hotel types shows what percentage of bookings belong to each hotel category.

### Real-World Example

A hotel company can use percentage distribution to understand the share of bookings received by each hotel type.

### AI/ML Usage

Percentage distribution helps identify dominant categories and understand whether a categorical variable is balanced or highly concentrated.

### Implementation

```python
percentage = df["hotel"].value_counts(normalize=True) * 100

print(percentage.round(2))
```

### Interpretation

The output shows the percentage of total bookings represented by each hotel type.

### Insight

The percentage distribution makes it easier to compare categories regardless of the total number of records.

### AI/ML Relevance

Understanding category proportions helps identify class imbalance and supports better decisions during data preprocessing and Machine Learning.

## 6. Categorical Visualizations

### 6.1 Bar Chart

A bar chart represents categorical data using rectangular bars. The size of each bar shows the value or frequency of a category.

### Example

A bar chart can be used to compare the number of bookings for different hotel types.

### Real-World Example

A hotel company can compare booking volumes between City Hotel and Resort Hotel.

### AI/ML Usage

Bar charts help compare categories and identify differences or dominant groups during Exploratory Data Analysis.

### Implementation

```python
plt.figure(figsize=(7, 4))

df["hotel"].value_counts().plot(kind="bar")

plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.show()
```

### Interpretation

The bar chart compares the number of bookings for each hotel type.

### Insight

The chart makes it easy to identify which hotel type has a higher number of bookings in the dataset.

### AI/ML Relevance

Bar charts help understand categorical feature distributions and support feature analysis before Machine Learning.

### 6.2 Count Plot

A count plot is a bar chart that shows the number of records in each category.

### Example

A count plot can be used to show the number of bookings for each hotel type.

### Real-World Example

A hotel company can use a count plot to compare the number of bookings received by different hotel types.

### AI/ML Usage

Count plots help identify category frequencies and reveal whether some categories are much more common than others.

### Implementation

```python
plt.figure(figsize=(7, 4))

sns.countplot(data=df, x="hotel")

plt.title("Hotel Type Booking Count")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.show()
```

### Interpretation

The count plot shows the number of bookings for each hotel type.

### Insight

The difference in bar heights indicates which hotel type has more booking records in the dataset.

### AI/ML Relevance

Count plots help identify category imbalance and understand categorical features before Machine Learning preprocessing.

### 6.3 Pie Chart

A pie chart represents the proportion of each category as a part of the whole.

### Example

A pie chart can be used to show the percentage distribution of bookings between different hotel types.

### Real-World Example

A hotel company can use a pie chart to understand the overall share of bookings between City Hotel and Resort Hotel.

### AI/ML Usage

Pie charts are useful for presenting the proportion of low-cardinality categorical variables and understanding category distribution.

### Implementation

```python
hotel_counts = df["hotel"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(hotel_counts, labels=hotel_counts.index, autopct="%1.1f%%")
plt.title("Hotel Type Booking Distribution")
plt.show()
```

### Interpretation

The pie chart shows the percentage share of bookings for each hotel type.

### Insight

The chart makes it easy to compare the contribution of each hotel category to the total number of bookings.

### AI/ML Relevance

Pie charts can help communicate categorical distributions clearly, especially when the number of categories is small.
