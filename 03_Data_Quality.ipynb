# Module 3 - Data Quality Assessment

This module focuses on identifying and quantifying data-quality issues in the Hotel Booking Demand dataset before performing data cleaning or preprocessing.


### A.1 Number of Missing Values

Missing values are records where information is not available or has not been recorded.

The total number of missing values helps us understand how much incomplete data is present in the dataset.

### Real-World Example

A hotel booking record may have missing information such as the number of children, agent, company, or other booking details.

### Implementation

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
total_missing = df.isnull().sum().sum()

print("Total number of missing values:", total_missing)
```

### Interpretation

The output shows the total number of missing values across all columns in the Hotel Booking Demand dataset.

### Insight

This gives an initial understanding of the overall data-quality condition and indicates whether missing values need further investigation.

### AI/ML Relevance

Identifying missing values is an important step before Machine Learning because incomplete data may affect analysis and model performance.

### A.2 Percentage of Missing Values by Column

The percentage of missing values shows how much data is missing in each column relative to the total number of records.

This helps identify columns where missing values may have a greater impact on the analysis.

### Example

A column with 100 missing values in a dataset of 10,000 records has 1% missing data.

### Real-World Example

In a hotel booking dataset, some booking-related columns may contain more missing values than others. Comparing percentages helps identify which columns need closer investigation.

### Implementation

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

missing_summary = missing_summary[missing_summary["Missing Count"] > 0]

print(missing_summary.sort_values("Missing Count", ascending=False))
```

### Interpretation

The output shows the number and percentage of missing values in each affected column.

### Insight

The missing-value percentage helps compare the severity of missing data across columns, even when the columns contain different amounts of data.

### AI/ML Relevance

Understanding the percentage of missing values helps determine which features may require further investigation before Machine Learning preprocessing.

### A.3 Missing Values by Column

Missing values by column show which specific columns contain incomplete data and how many values are missing in each column.

This helps identify the columns that require further investigation during data quality assessment.

### Example

A hotel booking dataset may have missing values in columns related to agents, companies, or guest information.

### Real-World Example

A hotel management team can identify which booking details are frequently missing and investigate the possible reasons.

### Implementation

```python
missing_by_column = df.isnull().sum()

print(missing_by_column[missing_by_column > 0])
```

### Interpretation

The output lists only the columns that contain missing values and shows the number of missing records in each column.

### Insight

This makes it easier to identify which specific features have data-quality issues.

### AI/ML Relevance

Knowing which features contain missing values helps determine which columns may need special treatment during later preprocessing.

### A.4 Missing Value Patterns

Missing value patterns help us understand how missing values are distributed across rows and columns.

It is useful to check how many rows contain missing values and whether multiple missing values appear together in the same records.

### Example

Some hotel booking records may have one missing value, while other records may have missing values in more than one column.

### Real-World Example

A hotel can investigate whether incomplete booking records are isolated cases or whether missing information commonly occurs across multiple booking fields.

### Implementation

```python
rows_with_missing = df.isnull().any(axis=1).sum()
missing_per_row = df.isnull().sum(axis=1)

print("Rows with missing values:", rows_with_missing)
print("\nMissing values per row:")
print(missing_per_row.value_counts().sort_index())
```

### Interpretation

The output shows how many rows contain at least one missing value and how missing values are distributed across individual rows.

### Insight

This helps determine whether missing data is concentrated in a small number of records or spread across many records.

### AI/ML Relevance

Understanding missing-value patterns helps choose suitable preprocessing strategies and prevents important data-quality issues from being overlooked before Machine Learning.

### B.1 Identify Duplicates

Duplicate records are rows that contain the same values across the relevant columns.

Identifying duplicates helps determine whether the dataset contains repeated records that may affect the accuracy of analysis.

### Example

The same hotel booking record may appear more than once due to repeated data entry or data collection issues.

### Real-World Example

A hotel system may accidentally store the same booking record multiple times, which can affect booking counts and business analysis.

### Implementation

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate rows:", len(duplicate_rows))
print(duplicate_rows.head())
```

### Interpretation

The output shows the records that have exact duplicate values in the dataset.

### Insight

Duplicate records can lead to incorrect counts and biased analysis if they are not identified.

### AI/ML Relevance

Duplicate records may cause certain observations to be overrepresented and can affect Machine Learning model training and evaluation.

### B.2 Count Duplicates

Counting duplicate records helps determine how many repeated rows are present in the dataset.

It also helps measure how much of the dataset is affected by duplication.

### Example

If the same hotel booking appears twice, one of those records is a duplicate.

### Real-World Example

Duplicate booking records can lead to incorrect booking counts and affect business reports.

### Implementation

```python
duplicate_count = df.duplicated().sum()
duplicate_percentage = (duplicate_count / len(df) * 100).round(2)

print("Duplicate Rows:", duplicate_count)
print("Duplicate Percentage:", duplicate_percentage, "%")
```

### Interpretation

The output shows the total number of exact duplicate rows and their percentage of the dataset.

### Insight

The duplicate percentage indicates how much of the dataset contains repeated records.

### AI/ML Relevance

Duplicate records can give repeated observations more influence during model training and may lead to misleading analysis or model performance.

### B.3 Identify Completely Duplicated Rows

A completely duplicated row is a record where all column values are exactly the same as another record.

Identifying complete duplicates helps determine whether the dataset contains repeated records.

### Example

If two hotel booking records have exactly the same values across every column, they are completely duplicated rows.

### Real-World Example

Duplicate booking records may occur because of repeated data entry or repeated data collection.

### Implementation

```python
duplicate_mask = df.duplicated(keep=False)

print("Rows involved in complete duplicates:", duplicate_mask.sum())
print(df[duplicate_mask].head())
```

### Interpretation

The output shows the records where all column values are repeated.

### Insight

Completely duplicated rows may cause the same booking information to be counted more than once during analysis.

### AI/ML Relevance

Repeated records can give certain observations extra weight and may affect Machine Learning results if they are not properly identified.

### B.4 Identify Potential Duplicate Records

Potential duplicate records are records that may represent the same real-world booking even when all column values are not exactly the same.

A common way to investigate potential duplicates is to check whether important identifying fields appear more than once.

### Example

If the same booking is recorded multiple times with slightly different values, it may not appear as a completely duplicated row but may still represent the same record.

### Real-World Example

A hotel booking system may contain repeated booking references because of updates, corrections, or repeated data entry.

### Implementation

```python
potential_duplicates = df[df.duplicated(subset=["arrival_date_year",
                                                 "arrival_date_month",
                                                 "arrival_date_day_of_month",
                                                 "lead_time",
                                                 "adults",
                                                 "children",
                                                 "babies",
                                                 "hotel"], keep=False)]

print("Potential duplicate rows:", len(potential_duplicates))
print(potential_duplicates.head())
```

### Interpretation

The output shows records that share the same selected booking-related details.

### Insight

Potential duplicates require further investigation because matching values in selected columns do not always mean the records are true duplicates.

### AI/ML Relevance

Identifying potential duplicates helps reduce repeated or inconsistent observations that could affect data analysis and Machine Learning results.

### C.1 Negative Values Where Inappropriate

Negative values should be checked in numerical columns where negative values do not make logical sense.

For example, values such as the number of guests, nights stayed, or room charges should not normally be negative.

### Example

A booking record with a negative number of adults or a negative room charge would indicate a possible data-quality issue.

### Real-World Example

A hotel cannot normally have a booking with -2 adults or a negative room rate. Such values may result from data-entry or system errors.

### Implementation

```python
numeric_columns = df.select_dtypes(include="number").columns

negative_counts = (df[numeric_columns] < 0).sum()

print("Negative values by column:")
print(negative_counts[negative_counts > 0])
```

### Interpretation

The output shows which numerical columns contain negative values and the number of such records.

### Insight

Negative values in columns where they are logically impossible should be investigated as potential data-quality issues.

### AI/ML Relevance

Invalid numerical values can affect statistical analysis and Machine Learning models, so they should be identified before preprocessing.

### C.2 Impossible / Out-of-Range Values

Some values may be positive but still fall outside a reasonable range for the dataset.

These values should be checked based on the meaning of each column and the business context.

### Example

A hotel booking record with an unusually large number of adults or an unrealistic stay duration may indicate a possible data-quality issue.

### Real-World Example

A hotel may have practical limits on the number of guests, nights stayed, or rooms booked in a single reservation.

### Implementation

```python
checks = {
    "adults > 10": (df["adults"] > 10).sum(),
    "children > 10": (df["children"] > 10).sum(),
    "babies > 5": (df["babies"] > 5).sum(),
    "total nights > 60": (
        (df["stays_in_weekend_nights"] + df["stays_in_week_nights"]) > 60
    ).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

### Interpretation

The output shows how many records fall outside the selected reasonable ranges.

### Insight

Out-of-range values should be investigated individually because an unusual value may be either a data error or a genuine exceptional booking.

### AI/ML Relevance

Impossible or unrealistic values can distort statistical analysis and influence Machine Learning models, so they should be identified before preprocessing.

### C.3 Incorrect / Inconsistent Categories

Categorical columns should contain values that follow a consistent set of categories.

Different spellings, capitalization, or unexpected labels can indicate data-quality issues.

### Example

A hotel booking column may contain values such as `Male`, `male`, and `M`, even though they represent the same category.

### Real-World Example

Inconsistent category values can occur when information is entered manually or collected from different systems.

### Implementation

```python
categorical_columns = df.select_dtypes(include="str").columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

### Insight

Unexpected spellings, different capitalization, or inconsistent labels can be identified by reviewing the category values.

### AI/ML Relevance

Inconsistent categorical values can create unnecessary categories and affect encoding and Machine Learning model performance.

### C.7 Empty Strings

Empty strings are text values that contain no characters.

They are different from `NaN` values and may not be detected by `isnull()`.

### Example

A categorical field in a hotel booking dataset may contain an empty value instead of a proper category.

### Real-World Example

An empty booking field may occur when information was not entered or was not captured correctly.

### Implementation

```python
empty_strings = {}

for col in df.select_dtypes(include="str").columns:
    count = (df[col].str.strip() == "").sum()
    if count > 0:
        empty_strings[col] = count

print("Empty strings:", empty_strings if empty_strings else "None found")
```

### Interpretation

The output shows whether any text columns contain empty string values.

### Insight

Empty strings can hide incomplete records and should be identified separately from standard missing values.

### AI/ML Relevance

Undetected empty strings can create incorrect categories during preprocessing and affect Machine Learning results.

### C.8 Null-like Values

Null-like values are text values that represent missing information but are not stored as actual missing values such as NaN.

Common examples include `NA`, `N/A`, `null`, `None`, `-`, `?`, and `Unknown`.

### Example

A hotel booking dataset may contain a value such as `Unknown` or `N/A` instead of a properly recorded category.

### Real-World Example

A booking system may store missing customer or booking information using placeholder text instead of leaving the value empty.

### Implementation

```python
null_like_values = ["na", "n/a", "null", "none", "-", "?", "unknown"]

found = {}

for col in df.select_dtypes(include="str").columns:
    mask = df[col].str.strip().str.lower().isin(null_like_values)
    if mask.sum() > 0:
        found[col] = mask.sum()

print("Null-like values:", found if found else "None found")
```

### Interpretation

The output shows whether any text columns contain placeholder values that may represent missing information.

### Insight

Checking for null-like values is important because these values may not be detected by `isnull()` and can remain hidden during data inspection.

### AI/ML Relevance

Identifying null-like values helps ensure that missing information is correctly recognized before data preprocessing and Machine Learning.
