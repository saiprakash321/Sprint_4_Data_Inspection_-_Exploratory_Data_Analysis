# Module 14 - EDA Mini Challenge

## 1. Load the Dataset

Load the dataset and check whether it is loaded correctly.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

## 2. Understand the Dataset

In this step, we get a basic understanding of the dataset by looking at the first few records and the number of rows and columns.

```python
print("First 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)
```

## 3. Inspect the Structure

In this step, we check the structure of the dataset and understand the columns, number of entries, and data types.

```python
print("Dataset Information:")
df.info()
```

## 4. Identify Data Types

In this step, we check the data type of each column and identify numerical, categorical, and other types of data.

```python
print("Data Types:")
display(df.dtypes)
```

## 5. Identify Missing Values

In this step, we check for missing values in each column and understand where data is incomplete.

```python
print("Missing Values:")
display(df.isnull().sum())

print("\nMissing Value Percentage:")
display((df.isnull().mean() * 100).sort_values(ascending=False))
```

## 6. Identify Duplicates

In this step, we check whether the dataset contains duplicate records.

```python
print("Number of Duplicate Records:", df.duplicated().sum())
```

## 7. Analyze Numerical Features

In this step, we identify the numerical columns and check their basic statistical values.

```python
print("Numerical Features:")
print(df.select_dtypes(include="number").columns.tolist())

print("\nStatistical Summary:")
display(df.select_dtypes(include="number").describe())
```

## 8. Analyze Categorical Features

In this step, we identify the categorical columns and check the different categories and their counts.

```python
categorical_cols = df.select_dtypes(include="object").columns

print("Categorical Features:")
print(categorical_cols.tolist())

for col in categorical_cols:
    print(f"\n{col}:")
    display(df[col].value_counts())
```

## 9. Detect Outliers

In this step, we check the numerical features for unusual values using the IQR method.

```python
num_df = df.select_dtypes(include="number")

Q1 = num_df.quantile(0.25)
Q3 = num_df.quantile(0.75)
IQR = Q3 - Q1

outliers = ((num_df < (Q1 - 1.5 * IQR)) |
            (num_df > (Q3 + 1.5 * IQR))).sum()

print("Number of Outliers:")
display(outliers.sort_values(ascending=False))
```

## 10. Analyze Distributions

In this step, we analyze the distribution of numerical features to understand their shape and identify any skewness.

```python
# Plot distributions of numerical features
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

## 11. Analyze Correlations

In this step, we check the relationship between numerical features and identify features that are strongly or weakly related.

```python
# Calculate correlation between numerical features
correlation = df.select_dtypes(include="number").corr()

display(correlation)
```

```python
# Visualize the correlations
plt.figure(figsize=(12, 8))

sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0)

plt.title("Correlation Heatmap")
plt.show()
```

## 12. Create Meaningful Visualizations

In this step, we create useful visualizations to understand the important patterns and relationships in the dataset.

```python
# Numerical feature distributions
num_cols = df.select_dtypes(include="number").columns

for col in num_cols[:5]:
    plt.figure(figsize=(8, 4))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# Categorical feature analysis
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols[:3]:
    plt.figure(figsize=(10, 5))
    df[col].value_counts().head(10).plot(kind="bar")
    plt.title(f"Top Categories in {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

## 13. Identify Important Patterns

In this step, we look at the results and visualizations to identify important patterns, trends, relationships, and unusual observations in the data.

```python
# Check the main numerical relationships
display(df.select_dtypes(include="number").corr().round(2))

# Check the most common values in categorical features
for col in df.select_dtypes(include="object").columns[:5]:
    print(f"\n{col}")
    display(df[col].value_counts().head(10))
```
