# Module 10 - Target Variable Analysis

Target variable analysis focuses on understanding the variable that the Machine Learning model is expected to predict.

In the Hotel Booking Demand dataset, `is_canceled` is the target variable.

## 1. Setup and Data Load

The required libraries are imported and the Hotel Booking Demand dataset is loaded for target variable analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 2. Class Distribution

Class distribution shows how the target variable is divided into its different classes.

In the Hotel Booking Demand dataset, `is_canceled` is the target variable, where `0` represents a booking that was not canceled and `1` represents a canceled booking.

### Implementation

```python
target_distribution = df["is_canceled"].value_counts().sort_index()

print("Class Distribution:")
print(target_distribution)

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="is_canceled")

plt.title("Booking Cancellation Class Distribution")
plt.xlabel("Is Canceled")
plt.ylabel("Number of Bookings")
plt.xticks([0, 1], ["Not Canceled", "Canceled"])
plt.show()
```

### Interpretation

The count plot shows the number of canceled and non-canceled bookings.

### AI/ML Relevance

The visualization helps identify whether the target classes are balanced or imbalanced before training a classification model.

## 3. Class Counts

Class counts show the total number of records belonging to each class of the target variable.

In the Hotel Booking Demand dataset, `0` represents bookings that were not canceled and `1` represents bookings that were canceled.

### Implementation

```python
class_counts = df["is_canceled"].value_counts().sort_index()

print("Class Counts:")
print("Not Canceled (0):", class_counts.get(0, 0))
print("Canceled (1):", class_counts.get(1, 0))
```

```python
plt.figure(figsize=(7, 4))

class_counts.plot(kind="bar")

plt.title("Class Counts of Booking Status")
plt.xlabel("Is Canceled")
plt.ylabel("Number of Bookings")
plt.xticks([0, 1], ["Not Canceled", "Canceled"], rotation=0)
plt.tight_layout()
plt.show()
```

### Interpretation

The output and bar chart show the number of bookings in each target class.

### AI/ML Relevance

Class counts help us understand how many observations are available for each class before training the classification model.

## 4. Class Percentages

Class percentages show the proportion of each class in the target variable.

### Implementation

```python
class_percentages = df["is_canceled"].value_counts(normalize=True).sort_index() * 100

print("Class Percentages:")
print("Not Canceled (0):", round(class_percentages.get(0, 0), 2), "%")
print("Canceled (1):", round(class_percentages.get(1, 0), 2), "%")
```

```python
plt.figure(figsize=(7, 4))

plt.bar(
    ["Not Canceled", "Canceled"],
    [
        class_percentages.get(0, 0),
        class_percentages.get(1, 0)
    ]
)

plt.title("Booking Cancellation Percentage")
plt.xlabel("Booking Status")
plt.ylabel("Percentage")
plt.ylim(0, 100)
plt.show()
```

### Interpretation

The output and bar chart show the percentage of bookings that were canceled and not canceled.

### AI/ML Relevance

Class percentages help identify whether the target variable has a balanced or imbalanced distribution before building a classification model.

## 5. Class Imbalance

Class imbalance occurs when one target class has more observations than another class.

### Implementation

```python
class_counts = df["is_canceled"].value_counts()

majority_class = class_counts.max()
minority_class = class_counts.min()

imbalance_ratio = majority_class / minority_class

print("Majority Class Count:", majority_class)
print("Minority Class Count:", minority_class)
print("Imbalance Ratio:", round(imbalance_ratio, 2))
```

```python
plt.figure(figsize=(7, 4))

sns.countplot(data=df, x="is_canceled")

plt.title("Class Imbalance in Booking Cancellation")
plt.xlabel("Is Canceled")
plt.ylabel("Number of Bookings")
plt.xticks([0, 1], ["Not Canceled", "Canceled"])
plt.show()
```

### Interpretation

The output and count plot show the difference in the number of observations between the two classes.

A larger difference indicates greater class imbalance.

### AI/ML Relevance

Class imbalance should be considered during model training and evaluation because a model may perform well on the majority class while performing poorly on the minority class.

## 6. Importance of Target Variable

The target variable represents the outcome that the Machine Learning model is expected to predict.

In the Hotel Booking Demand dataset, `is_canceled` is the target variable.

Understanding its class distribution and imbalance helps us prepare the data correctly before building a classification model.

### AI/ML Relevance

Target variable analysis helps determine the classification problem, understand class balance, and choose suitable model evaluation methods.
