# Module 5 - Bivariate Analysis

Bivariate analysis studies the relationship between two variables to understand how they are related and identify meaningful patterns in the dataset.

This notebook uses the Hotel Booking Demand dataset to analyze relationships between numerical and categorical variables using statistical methods and visualizations.

## 1. Setup and Data Load

The required Python libraries are imported and the Hotel Booking Demand dataset is loaded for bivariate analysis.

The dataset will be used to study relationships between numerical and categorical variables.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 2. Numerical vs Numerical

Numerical vs Numerical analysis studies the relationship between two numerical variables.

In the Hotel Booking Demand dataset, lead_time and adr can be used to study whether booking time is related to the average daily room rate.

### 2.1 Scatter Plot

A scatter plot is used to visualize the relationship between two numerical variables.

### Example

The relationship between lead time and average daily rate can help us understand how room rates vary with booking time.

### Real-World Example

A hotel can analyze whether customers who book earlier tend to have different room rates.

### AI/ML Usage

Scatter plots help identify relationships, trends, clusters, and unusual observations between numerical features.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.scatterplot(data=df, x="lead_time", y="adr")

plt.title("Lead Time vs Average Daily Rate")
plt.xlabel("Lead Time")
plt.ylabel("Average Daily Rate")
plt.show()

correlation = df[["lead_time", "adr"]].corr().iloc[0, 1]
print("Correlation:", round(correlation, 2))
```

### Interpretation

The scatter plot shows how the average daily rate varies with different booking lead times.

The correlation value indicates the strength and direction of the linear relationship between the two variables.

### Insight

The distribution of points helps identify whether a clear relationship exists between booking lead time and average daily rate.

### AI/ML Relevance

Understanding relationships between numerical features helps identify useful variables and possible correlations before Machine Learning.

## 3. Numerical vs Categorical

Numerical vs Categorical analysis studies how a numerical variable differs across different categories.

In the Hotel Booking Demand dataset, adr and hotel can be used to compare average daily rates across hotel types.

### 3.1 Box Plot

A box plot compares the distribution of a numerical variable across categories and shows the median, quartiles, spread, and potential outliers.

### Example

The distribution of average daily rate can be compared between City Hotel and Resort Hotel.

### Real-World Example

A hotel company can compare room-rate patterns between different hotel types.

### AI/ML Usage

Box plots help identify differences in distributions and potential outliers across categories.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="hotel", y="adr")

plt.title("Average Daily Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")
plt.show()
```

### Interpretation

The box plot compares the spread of average daily rates for each hotel type.

### Insight

The differences in the median and spread help identify how room rates vary between hotel categories.

### AI/ML Relevance

Comparing numerical features across categories helps identify useful patterns for feature analysis and Machine Learning.

### 3.2 Violin Plot

A violin plot shows the distribution and density of a numerical variable across different categories.

### Example

The distribution of average daily rate can be compared across hotel types using a violin plot.

### Real-World Example

A hotel company can use this visualization to understand where room prices are concentrated for different hotel types.

### AI/ML Usage

Violin plots help compare distributions, concentration, spread, and possible differences between categories.

### Implementation

```python
plt.figure(figsize=(8, 4))

sns.violinplot(data=df, x="hotel", y="adr")

plt.title("Average Daily Rate Distribution by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")
plt.show()
```

### Interpretation

The violin plot shows the distribution and density of average daily rates for each hotel type.

### Insight

The shape and width of each violin help identify where values are concentrated and how widely the rates are distributed.

### AI/ML Relevance

Distribution comparisons help understand how categorical variables are related to numerical features before model development.

## 4. Categorical vs Categorical

Categorical vs Categorical analysis studies the relationship between two categorical variables.

In the Hotel Booking Demand dataset, hotel and meal can be used to examine how meal preferences vary across hotel types.

### 4.1 Cross Tabulation

Cross tabulation summarizes the frequency of combinations of two categorical variables.

### Example

A cross tabulation can show how different meal types are distributed across City Hotel and Resort Hotel.

### Real-World Example

A hotel can use this information to understand meal preferences for different hotel types.

### AI/ML Usage

Cross tabulation helps identify relationships and category patterns between categorical features.

### Implementation

```python
cross_tab = pd.crosstab(df["hotel"], df["meal"])

print(cross_tab)
```

### Interpretation

The output shows the number of bookings for each combination of hotel type and meal type.

### Insight

The table helps identify which meal categories are more common for each hotel type.

### AI/ML Relevance

Cross tabulation helps understand relationships between categorical features and can support feature selection and encoding decisions.

### 4.2 Grouped Bar Chart

A grouped bar chart compares the frequency or value of multiple categories across groups.

### Example

A grouped bar chart can visually compare meal preferences across different hotel types.

### Real-World Example

Hotel management can use the chart to compare meal preferences between City Hotel and Resort Hotel.

### AI/ML Usage

Grouped bar charts make categorical relationships easier to compare and interpret during EDA.

### Implementation

```python
cross_tab.plot(kind="bar", figsize=(8, 4))

plt.title("Meal Distribution by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.legend(title="Meal")
plt.show()
```

### Interpretation

The grouped bar chart compares the frequency of different meal types across hotel categories.

### Insight

The chart makes it easier to identify differences in meal preferences between hotel types.

### AI/ML Relevance

Visual comparison of categorical features helps identify patterns that may be useful during feature analysis and Machine Learning preprocessing.
