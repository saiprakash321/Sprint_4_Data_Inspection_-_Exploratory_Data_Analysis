## 3. What is a Target?

A target is the variable or outcome that we want to predict or understand using the other variables in a dataset.

In Machine Learning, the target is also called the dependent variable or output variable.

### Example

In the Hotel Booking Demand dataset, the booking cancellation status is the target.

The target tells us whether a booking was canceled or not.

### Real-World Example

A hotel may want to predict whether a future booking is likely to be canceled based on booking and customer information.

### Implementation

```python
target = "is_canceled"

print("Target Variable:", target)
print(df[target].head())
```

### Output

The output shows the target column and its first few values.

### Interpretation

The target variable contains the outcome that we want to analyze or predict.

### Insight

The target helps define the main objective of the analysis and determines what the other features will be used to understand or predict.

### AI/ML Relevance

Identifying the target variable is an important step in Machine Learning because it defines the outcome that the model is expected to predict.
