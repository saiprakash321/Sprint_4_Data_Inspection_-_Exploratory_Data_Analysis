# Module 11 - EDA Visualization

EDA visualization is used to understand patterns, relationships, distributions, and important insights in the Hotel Booking Demand dataset.



## 1. Setup and Data Load

The required Python libraries are imported and the Hotel Booking Demand dataset is loaded for Exploratory Data Analysis visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/hotel_bookings.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 2. Histogram

A histogram is used to show the distribution of a numerical variable.

### Why use it?

It helps us understand where most values are concentrated and identify the overall distribution shape and unusual values.

### Implementation

```python
plt.figure(figsize=(8, 5))

plt.hist(df["lead_time"].dropna(), bins=20)

plt.title("Distribution of Lead Time")
plt.xlabel("Lead Time")
plt.ylabel("Number of Bookings")

plt.show()
```

### Interpretation

The histogram shows how hotel bookings are distributed across different lead-time ranges.

### Insight

The visualization helps identify the ranges where bookings are more concentrated and whether the lead-time distribution is skewed or contains unusual values.

## 3. Box Plot

A box plot is used to show the spread of a numerical variable and identify potential outliers.

### Why use it?

It helps us understand the median, quartiles, overall spread, and unusual values in the data.

### Implementation

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["adr"])

plt.title("Box Plot of Average Daily Rate")
plt.xlabel("Average Daily Rate")

plt.show()
```

### Interpretation

The box plot shows the distribution and spread of the average daily rate.

### Insight

The plot helps identify the typical range of room rates and any potential outliers that may need further investigation.

## 4. Bar Chart

A bar chart is used to compare the frequency of different categories.

### Why use it?

It helps compare the number of bookings across different hotel types.

### Implementation

```python
plt.figure(figsize=(8, 5))

hotel_counts = df["hotel"].value_counts()

plt.bar(hotel_counts.index, hotel_counts.values)

plt.title("Number of Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.show()
```

### Interpretation

The bar chart compares the number of bookings for each hotel type.

### Insight

The visualization helps identify which hotel type has more bookings in the dataset.

## 5. Count Plot

A count plot is used to show the number of observations in each category.

### Why use it?

It helps us understand the distribution of bookings based on the target variable `is_canceled`.

### Implementation

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="is_canceled")

plt.title("Booking Cancellation Distribution")
plt.xlabel("Is Canceled")
plt.ylabel("Number of Bookings")
plt.xticks([0, 1], ["Not Canceled", "Canceled"])

plt.show()
```

### Interpretation

The count plot shows the number of canceled and non-canceled bookings.

### Insight

The visualization helps identify the target class distribution and whether there is any class imbalance in the dataset.

## 6. Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

### Why use it?

It helps identify relationships, patterns, trends, and possible outliers between two numerical features.

### Implementation

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="lead_time",
    y="adr"
)

plt.title("Lead Time vs Average Daily Rate")
plt.xlabel("Lead Time")
plt.ylabel("Average Daily Rate")

plt.show()
```

### Interpretation

The scatter plot shows how the average daily rate varies with different lead times.

### Insight

The distribution of points helps us understand whether there is a visible relationship or pattern between lead time and average daily rate.

## 7. Line Plot

A line plot is used to show trends or changes across an ordered variable.

### Why use it?

The Hotel Booking Demand dataset is not a true time-series dataset, so a line plot should only be used with an ordered variable.

Here, we can use `lead_time` to observe how the average daily rate changes with booking lead time.

### Implementation

```python
lead_time_adr = df.groupby("lead_time")["adr"].mean()

plt.figure(figsize=(8, 5))

plt.plot(lead_time_adr.index, lead_time_adr.values)

plt.title("Average Daily Rate by Lead Time")
plt.xlabel("Lead Time")
plt.ylabel("Average Daily Rate")

plt.show()
```

### Interpretation

The line plot shows how the average daily rate changes across different lead-time values.

### Insight

The visualization helps identify any trend or variation in average daily rates as the booking lead time increases.

## 8. Violin Plot

A violin plot is used to show the distribution of a numerical variable across categories.

### Why use it?

It helps compare the distribution, spread, and density of average daily rates between different hotel types.

### Implementation

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="hotel",
    y="adr"
)

plt.title("Average Daily Rate Distribution by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")

plt.show()
```

### Interpretation

The violin plot shows how the average daily rate is distributed across the different hotel types.

### Insight

The shape and width of each violin help compare the concentration and spread of room rates between hotel types.

## 9. Pair Plot

A pair plot is used to show the relationships between multiple numerical variables at the same time.

### Why use it?

It helps identify relationships, patterns, clusters, and possible outliers among selected numerical features.

### Implementation

```python
numerical_features = [
    "lead_time",
    "adr",
    "stays_in_week_nights",
    "adults"
]

sns.pairplot(df[numerical_features].dropna())

plt.show()
```

### Interpretation

The pair plot shows the relationships between the selected numerical features, with individual distributions shown along the diagonal.

### Insight

The visualization provides an overall view of the relationships among the selected features and helps identify patterns that may need further analysis.

## 10. Heatmap

A heatmap is used to show the correlation between multiple numerical variables using different levels of visual intensity.

### Why use it?

It helps quickly identify strong, weak, and negative relationships between numerical features.

### Implementation

```python
numerical_features = [
    "lead_time",
    "adr",
    "stays_in_week_nights",
    "adults"
]

correlation_matrix = df[numerical_features].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

The heatmap displays the correlation values between the selected numerical features.

Values closer to +1 indicate a positive relationship, values closer to -1 indicate a negative relationship, and values close to 0 indicate a weak relationship.

### Insight

The heatmap helps identify important feature relationships and possible highly correlated variables that may require further analysis.

## 11. Distribution Plot

A distribution plot is used to understand how the values of a numerical variable are distributed.

### Why use it?

It helps identify the center, spread, shape, skewness, and possible unusual values in a numerical feature.

### Implementation

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="adr",
    kde=True
)

plt.title("Distribution of Average Daily Rate")
plt.xlabel("Average Daily Rate")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

The bars show the frequency of different average daily rate values, while the KDE curve provides a smooth view of the distribution.

### Insight

The distribution plot helps us understand the typical average daily rate, its spread, and whether the feature is approximately symmetric or skewed.

- The Box Plot helped identify the spread and potential outliers.

- The Scatter Plot helped identify relationships between numerical variables.
- The Line Plot showed changes across the ordered lead-time variable.
- The Violin Plot helped compare numerical distributions across hotel types.
- The Pair Plot provided an overall view of relationships among multiple numerical features.
