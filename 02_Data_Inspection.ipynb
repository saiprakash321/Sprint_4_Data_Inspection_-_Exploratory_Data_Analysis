## 1. Number of Rows

The number of rows represents the total number of records or observations present in a dataset.

In the Hotel Booking Demand dataset, each row represents one hotel booking record.

### Example

If a dataset contains 10,000 rows, it means there are 10,000 booking records available for analysis.

### Real-World Example

A hotel can use the number of records to understand how much booking history is available for analysis.

### Implementation

```python
import pandas as pd

df = pd.read_csv("data/hotel_bookings.csv")

df.head()
```

```python
print("Number of Rows:", df.shape[0])
```

### Output

The output shows the total number of hotel booking records in the dataset.

### Interpretation

The output represents the total number of booking records available for analysis.

### Insight

Knowing the number of rows helps us understand the size of the dataset before performing further analysis.

### AI/ML Relevance

The number of observations is important in Machine Learning because it indicates how much data is available for learning patterns and building reliable models.

## 15. Identify Important Column Types

After inspecting the dataset, we should identify numerical, categorical, date/time, identifier, and potential target columns.

### Numerical Columns

Numerical columns contain values that can be used for mathematical and statistical analysis.

### Categorical Columns

Categorical columns contain labels or groups.

### Date/Time Columns

Date/time columns contain information related to dates or booking time periods.

### Identifier Columns

Identifier columns are used to identify records or related entities and should be evaluated carefully before using them as predictive features.

### Potential Target Column

The potential target column is the variable that represents the outcome we want to understand or predict.

### Implementation

```python
numerical_columns = df.select_dtypes(include="number").columns.tolist()
categorical_columns = df.select_dtypes(include="str").columns.tolist()

date_columns = [
    col for col in df.columns
    if "date" in col.lower()
]

target_column = "is_canceled"

print("Numerical Columns:", numerical_columns)
print("Categorical Columns:", categorical_columns)
print("Date/Time Columns:", date_columns)
print("Potential Target:", target_column)
```

### Interpretation

The identified columns provide an initial classification of the dataset structure.

Numerical columns can be analyzed statistically, categorical columns can be analyzed by category, date-related columns can be examined for time-based patterns, and the target column represents the main outcome for analysis.

### Insight

This inspection helps us understand the dataset before moving to data-quality assessment and Exploratory Data Analysis.

### AI/ML Relevance

Correctly identifying feature types and the target variable is essential for choosing appropriate preprocessing, analysis, visualization, and Machine Learning techniques.
